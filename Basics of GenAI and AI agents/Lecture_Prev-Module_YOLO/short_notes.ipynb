- Anchor boxes
    - number of anchor boxes - hyperparameter
    - sizes - hyperparameter
    - each grid cell is associated with multiple anchor boxes

- 

- Anchor boxes are not the same as bounding boxes

- 

- During YOLO, we dont directly predict Xc, Yc, h and w , instead we predict, Tx,Ty, Tw, Th (Transformed values)

- Let bx, by, bw, bh be the values to be predicted
- Using transformed values, the bx, by, bw, bh are calculated as follows:
    - bx = σ(tx) + cx
    - by = σ(ty) + cy
        - cx, cy --> coordinates of the grid cell
    - bw = pw * e^tw
    - bh = ph * e^th
        - pw --> width of anchor box
        - ph --> height of anchor box

- YOLO wants each grid cell to predict objects whose centers lie within that cell, so
    - 
- cx < bx < cx+1
- cy < by < cy+1


- Derive tx from xc/bx during training
- Derive xc/bx from tx during inference
