## 2. The Temperature Parameter

The **temperature parameter** $T$ modifies the softmax function to control the "sharpness" of the probability distribution:

$$
\text{softmax}_T(z_i) = \frac{e^{z_i/T}}{\sum_{j=1}^{n} e^{z_j/T}}
$$

### Effects of Temperature

| Temperature | Effect | Use Case |
|-------------|--------|----------|
| $T \to 0$ | **Sharp/Peaked** distribution (approaches hard max) | Confident predictions, exploitation |
| $T = 1$ | **Standard** softmax | Default neural network outputs |
| $T > 1$ | **Smooth/Flat** distribution (more uniform) | Exploration, knowledge distillation |
| $T \to \infty$ | **Uniform** distribution | Maximum uncertainty |

### Intuition

- **Low temperature** ($T < 1$): Makes the model more confident and selective
- **High temperature** ($T > 1$): Makes the model less confident and more exploratory

## Summary

### Key Takeaways

1. **Temperature in softmax** controls the "sharpness" of probability distributions

2. **Low temperature** ($T < 1$):
   - Sharp, peaked distributions
   - High confidence predictions
   - Useful for: greedy decoding, deterministic outputs, exploitation

3. **High temperature** ($T > 1$):
   - Smooth, flat distributions
   - More uniform probabilities
   - Useful for: exploration, knowledge distillation, creative generation

4. **Applications**:
   - Language models: Control creativity in text generation
   - Knowledge distillation: Preserve class relationships
   - Reinforcement learning: Balance exploration vs exploitation
   - Calibration: Adjust model confidence

### Formula

$$
\text{softmax}_T(z_i) = \frac{e^{z_i/T}}{\sum_{j=1}^{n} e^{z_j/T}}
$$

where $T$ is the temperature parameter.

# LLM Generation & Inference Concepts

### 1. **Temperature**
Controls the randomness of token selection by scaling the probability distribution before sampling.  
- Lower = more deterministic outputs  
- Higher = more creative and diverse outputs

---

### 2. **do_sample**
Decides whether to sample randomly or always choose the highest-probability token.  
- **False** → Greedy/Beam Search (deterministic)  
- **True** → Top-k/Top-p sampling (stochastic)

---

### 3. **Top-k Sampling**
Restricts sampling to only the **k** most probable next tokens.  
Example: `top_k = 50` means every token outside the top 50 probabilities is ignored.

---

### 4. **Top-p (Nucleus Sampling)**
Samples from the smallest set of tokens whose cumulative probability is at least **p**.  
More adaptive than Top-k because the candidate set changes at every generation step.

---

### 5. **Greedy Decoding**
Always selects the highest-probability next token.  
- Fast and deterministic  
- May produce suboptimal overall sequences

---

### 6. **Beam Search**
Maintains multiple candidate sequences simultaneously and chooses the one with the highest overall probability.  
Commonly used for translation and summarization.

---

### 7. **Presence Penalty**
Penalizes tokens that have already appeared at least once.  
Encourages introducing new ideas and topics.

---

### 8. **Frequency Penalty**
Penalizes tokens based on how often they have already appeared.  
Helps reduce repetitive wording.

---

### 9. **Repetition Penalty**
Directly lowers the probability of previously generated tokens.  
Commonly used in Hugging Face models to reduce repetitive loops.

---

### 10. **max_new_tokens**
Maximum number of new tokens the model is allowed to generate.  
Does **not** include the prompt tokens.

---

### 11. **max_length**
Maximum total sequence length (prompt + generated output).  
Older APIs commonly used this parameter; newer APIs often recommend `max_new_tokens`.

---

### 12. **min_new_tokens**
Forces the model to generate at least a specified minimum number of tokens.  
Prevents very short answers.

---

### 13. **Stop Sequences (`stop`)**
Specifies strings where generation should immediately stop.  
Example: stop at `"User:"` or `"\n\n"`.

---

### 14. **EOS Token (End of Sequence)**
A special token indicating the model should stop generating.  
Usually generated automatically unless generation limits are reached first.

---

### 15. **BOS Token (Beginning of Sequence)**
A special token marking the start of the input sequence.  
Used internally during tokenization.

---

### 16. **PAD Token**
Added to shorter sequences so all inputs in a batch have equal length.  
Ignored during attention using an attention mask.

---

### 17. **Attention Mask**
Indicates which tokens are actual input and which are padding.  
Prevents the model from attending to padded positions.

---

### 18. **Context Window**
The maximum number of tokens the model can consider at one time.  
Larger context windows enable longer conversations and document processing.

---

### 19. **Token**
The smallest unit processed by an LLM (word, subword, or character fragment).  
Models generate one token at a time.

---

### 20. **Logits**
Raw scores produced by the model before applying Softmax.  
Decoding strategies operate directly on these logits.

---

### 21. **Softmax**
Converts logits into probabilities that sum to 1.  
The next token is selected from this probability distribution.

---

### 22. **Log Probabilities (`logprobs`)**
The natural logarithm of token probabilities.  
Used to estimate confidence and compare generated sequences.

---

### 23. **Seed**
Initializes the random number generator.  
Using the same prompt, model, seed, and parameters usually produces reproducible outputs.

---

### 24. **num_beams**
Specifies the number of beams used during Beam Search.  
Larger values improve search quality but increase inference time.

---

### 25. **num_return_sequences**
Determines how many different completions are returned.  
Often combined with Beam Search or sampling.

---

### 26. **Early Stopping**
Stops Beam Search when all candidate beams have finished.  
Reduces unnecessary computation.

---

### 27. **Length Penalty**
Adjusts Beam Search's preference for shorter or longer outputs.  
Helps avoid excessively short translations.

---

### 28. **Typical Sampling**
Samples tokens close to the distribution's expected information content.  
Attempts to balance coherence and diversity.

---

### 29. **Temperature Annealing**
Dynamically changes the temperature during generation.  
Typically starts high for exploration and lowers later for consistency.

---

### 30. **Streaming**
Returns generated tokens one at a time as they are produced.  
Improves perceived responsiveness in chat applications.

---

### 31. **Speculative Decoding**
Uses a smaller draft model to propose tokens and a larger model to verify them.  
Speeds up inference without changing the final output.

---

### 32. **KV Cache (Key-Value Cache)**
Stores attention keys and values from previous tokens.  
Avoids recomputing attention, significantly accelerating autoregressive generation.

---

### 33. **Batch Size (Inference)**
The number of prompts processed simultaneously during inference.  
Larger batches improve throughput but require more GPU memory.

---

### 34. **Sampling vs Decoding**
- **Sampling:** Randomly selects tokens from a probability distribution (e.g., Top-k, Top-p).  
- **Decoding:** Refers to the overall strategy used to generate tokens (e.g., Greedy, Beam Search, Sampling).

---

### 35. **Hallucination**
Occurs when the model generates fluent but incorrect or fabricated information.  
Can be reduced using RAG, better prompts, grounding, and verification.

---

### 36. **Deterministic vs Stochastic Generation**
- **Deterministic:** Same input always produces the same output (Greedy, Beam Search).  
- **Stochastic:** Output may differ due to randomness (Sampling).
