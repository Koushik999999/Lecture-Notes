### Pre Training

#### Pretraining objectives:
- Masked Language Modeling (MLM)
    - How words are used within a sentence 
- Next Sentence Prediction (NSP)


#### Masked Language Modeling (MLM)

- Across sentence context is not captured by Masked Language Modeling.
- So we need to train the model on next sentence prediction.

#### Next Sentence Prediction objective

- Capture the relationship between sentences

- It's a Binary classification 
- Log loss is used


- Input: 
    - [CLS] the [MASK] is big. It is [MASK] of forest 
- Output:
    - [CLS] : 1
    - [MASK] : Lion
    - [MASK] : King
- Model is trained to predict this output with the above input
