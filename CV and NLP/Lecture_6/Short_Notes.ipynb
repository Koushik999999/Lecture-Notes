- Write a function computeiou(box1, box2, imgw, img_h) that calculates the Intersection over Union score for two YOLO boxes.
- box1 → list of 4 floats: [x_center, y_center, width, height] (normalized between 0–1)
- box2 → list of 4 floats: [x_center, y_center, width, height] (normalized between 0–1)
- img_w → int, image width in pixels
- img_h → int, image height in pixels

```python
def calculate_iou(box1, box2, img_w, img_h):
    box1_xleft = (box1[0] - box1[2]/2)*img_w
    box2_xleft = (box2[0] - box2[2]/2)*img_w
    box1_ytop = (box1[1] - box1[3]/2)*img_h
    box2_ytop = (box2[1] - box2[3]/2)*img_h

    xleft = max(box1_xleft,box2_xleft)
    ytop = max(box1_ytop,box2_ytop)
    
    box1_ydown = (box1[1] + box1[3]/2)*img_h
    box2_ydown = (box2[1] + box2[3]/2)*img_h
    box1_xright = (box1[0] + box1[2]/2)*img_w
    box2_xright = (box2[0] + box2[2]/2)*img_w

    ydown = min(box1_ydown,box2_ydown)
    xright = min(box1_xright,box2_xright)
    
    
    intersection_width = xright - xleft
    intersection_height = ydown - ytop
    if(intersection_width < 0): intersection_width = 0
    if(intersection_height < 0): intersection_height = 0
    intersection_area = intersection_height*intersection_width
    box1_area = box1[2]*img_w*box1[3]*img_h
    box2_area = box2[2]*img_w*box2[3]*img_h
    union_area = box1_area + box2_area - intersection_area

    iou = intersection_area/union_area

    return round(iou,2)
```

```python
box1 = [0.5, 0.5, 0.4, 0.4]
box2 = [0.55, 0.55, 0.4, 0.4]
img_w = 416
img_h = 416

print(calculate_iou(box1,box2,img_w,img_h))
```
