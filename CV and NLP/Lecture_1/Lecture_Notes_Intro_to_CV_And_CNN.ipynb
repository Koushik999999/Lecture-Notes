```python
def mlops_timer(func):
    """A decorator for timing ML operations"""
    def wrapper(*args, **kwargs):
        func_name = func.__name__
        print(f"⏱️  Started {func_name}")
        
        start = time.time()
        result = func(*args, **kwargs)
        duration = time.time() - start
        
        # Smart formatting
        if duration < 1:
            duration_str = f"{duration*1000:.1f}ms"
        else:
            duration_str = f"{duration:.2f}s"
        
        print(f"✅ {func_name} completed in {duration_str}")
        return result
    return wrapper

# Apply to ALL our ML functions
@mlops_timer
def load_data(filepath):
    print("   Loading data...")
    time.sleep(0.5)
    return "data_loaded"

@mlops_timer
def preprocess(data):
    print("   Preprocessing...")
    time.sleep(0.3)
    return "data_preprocessed"

@mlops_timer
def train_model(data):
    print("   Training model...")
    time.sleep(2)
    return "model_trained"

@mlops_timer
def evaluate(model):
    print("   Evaluating...")
    time.sleep(0.8)
    return {"accuracy": 0.95}

# Run a complete pipeline
data = load_data("train.csv")
clean_data = preprocess(data)
model = train_model(clean_data)
metrics = evaluate(model)
```

```python
def log_level(level):              # Level 1: Configuration
    print(f"🔧 Level 1 executed with level='{level}'")
    
    def decorator(func):           # Level 2: Receives function
        print(f"🎨 Level 2 executed with func={func.__name__}")
        
        def wrapper(*args, **kwargs):  # Level 3: Wraps execution
            print(f"📝 [{level}] Running {func.__name__}...")  # Uses 'level' from Level 1!
            result = func(*args, **kwargs)
            print(f"✅ [{level}] {func.__name__} completed")
            return result
        
        return wrapper  # Return Level 3
    
    return decorator   # Return Level 2
```

```python
@log_level("DEBUG")
def train_model():
    print("   Training the model...")
    return "Model trained"

print("\n" + "="*50)
print("Now calling the function:")
print("="*50 + "\n")

result = train_model()
print(f"Result: {result}")
```

```python
import functools

def better_timer(func):
    """An improved timer decorator"""
    @functools.wraps(func)  # ← This is the magic!
    def wrapper(*args, **kwargs):
        print("Timing...")
        return func(*args, **kwargs)
    return wrapper
```

```python
@better_timer
def train_model(epochs, learning_rate=0.01):
    """
    Trains a machine learning model.
    
    Args:
        epochs: Number of training epochs
        learning_rate: Learning rate for optimization
        
    Returns:
        Trained model
    """
    print(f"Training for {epochs} epochs with lr={learning_rate}")
    return "Model trained"

# Now let's inspect again
print("Function name:", train_model.__name__)
print("Function docstring:", train_model.__doc__)
print("\n" + "="*50)
help(train_model)
```
