# Word Embeddings

- The vector representing word should also contain meaning


- CBoW
- Skip Gram

- CBoW

    - Predicts a word based on its context
    - Example:
        - "I ___ to school" → predicts "go"

- Skip-gram
    - Predicts context words from a given word
    - Example:
        - "go" → predicts ["I", "to", "school"]

### Usage:

```python
import gensim.downloader as api

# Load pre-trained Word2Vec (Google News-like smaller model)
model = api.load("word2vec-google-news-300")

# Get embedding of a word
vector = model["computer"]

print(vector)


## DonT run this --> 1.5 GB data 
```

```python
## How it is trained on a smaller data:

from gensim.models import Word2Vec

# Sample data
sentences = [
    ["i", "love", "nlp"],
    ["nlp", "is", "fun"],
    ["i", "love", "machine", "learning"]
]

# Train model
model1 = Word2Vec(sentences, vector_size=5, window=2, min_count=1)

# Convert word to embedding
vector = model1.wv["love"]

 # vec1 = model1.wv["koushik"]  --> gives error for words not present in vocabulary 
print(vector)
# print(vec1)
```
