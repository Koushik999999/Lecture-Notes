# Agentic Workflow Patterns – Interview Cheat Sheet

---

# 1. Agentic Workflow

## Definition

A structured pattern that allows an LLM to solve complex tasks using reasoning, tools, verification, and planning.

More reliable than a single prompt because work is divided into multiple stages.

---

# 2. Prompt Chaining

## Definition

Splits one large task into multiple sequential prompts.

Each prompt uses the output of the previous prompt.

### Pipeline

```text
Input
   │
   ▼
Prompt 1
   │
   ▼
Prompt 2
   │
   ▼
Prompt 3
   │
   ▼
Final Output
```

### Advantages

- Easier debugging
- Better accuracy
- Modular design

### Example

```text
Generate SQL
     │
     ▼
Execute SQL
     │
     ▼
Summarize Results
```

---

# 3. Routing

## Definition

Sends the input to the most appropriate prompt, tool, or specialized agent.

Only one route executes.

### Pipeline

```text
User Query
      │
      ▼
    Router
 ┌────┼────┐
 ▼    ▼    ▼
SQL Search Code
```

### Example

```text
SQL Question
     │
     ▼
SQL Agent

Weather Question
     │
     ▼
Weather API
```

### Advantages

- Faster
- More accurate
- Uses specialized expertise

---

# 4. Parallelization

## Definition

Multiple tasks execute simultaneously.

Results are combined afterward.

### Pipeline

```text
          Input
            │
     ┌──────┼──────┐
     ▼      ▼      ▼
 Task1   Task2   Task3
     │      │      │
     └──────┼──────┘
            ▼
    Combined Result
```

### Advantages

- Faster execution
- Independent verification
- Better scalability

---

# 5. Sectioning Pattern

## Definition

Splits a large input into smaller independent sections.

Each section is processed separately.

### Example

```text
100-page PDF
      │
      ▼
10 Sections
      │
      ▼
Analyze Separately
      │
      ▼
Merge Results
```

---

# 6. Voting Pattern

## Definition

Multiple agents independently solve the same problem.

Final answer is selected by majority vote or confidence.

### Pipeline

```text
Question
   │
┌──┼──┐
▼  ▼  ▼
LLM1
LLM2
LLM3
   │
   ▼
Majority Vote
   │
   ▼
Final Answer
```

### Advantages

- Reduces hallucinations
- Improves reasoning reliability

---

# 7. Evaluator–Optimizer Workflow

## Definition

One model generates an answer.

Another model evaluates and suggests improvements.

The process repeats until quality is acceptable.

### Pipeline

```text
Generate
    │
    ▼
Evaluate
    │
    ▼
Improve
    │
    ▼
Evaluate
    │
    ▼
Final Answer
```

---

# 8. Evaluator

## Definition

Reviews generated output against predefined criteria.

Identifies errors or missing information.

### Example

```text
Generated SQL
      │
      ▼
Syntax Check
      │
      ▼
Logical Check
      │
      ▼
Feedback
```

---

# 9. Optimizer

## Definition

Uses evaluator feedback to improve the output.

Continues refinement until requirements are met.

---

# 10. Evaluation Loop

## Definition

Continuous:

**Generate → Evaluate → Improve**

Stops when the quality threshold is reached.

---

# 11. Orchestrator–Worker Workflow

## Definition

One central orchestrator divides work among specialized workers.

Workers execute independently and return results.

### Pipeline

```text
User Request
      │
      ▼
Orchestrator
      │
      ▼
 ┌────┼────┐
 ▼    ▼    ▼
Worker A
Worker B
Worker C
      │
      ▼
Merge Results
      │
      ▼
Final Answer
```

---

# 12. Orchestrator

## Definition

Plans the task.

Assigns work to specialized workers.

Combines final outputs.

---

# 13. Worker Agent

## Definition

Performs one specialized task.

Reports results back to the orchestrator.

### Examples

- SQL Worker
- Search Worker
- Python Worker
- Summary Worker

---

# 14. SQL Generation Agent

## Definition

Converts natural language into SQL queries.

Uses schema information before generating SQL.

---

# 15. SQL Evaluator

## Definition

Validates generated SQL.

Detects syntax and logical errors before execution.

---

# 16. Execution Validator

## Definition

Executes generated SQL safely.

Verifies whether execution succeeded.

---

# 17. Repair Stage

## Definition

Automatically fixes failed SQL using error messages.

Common in Text-to-SQL agents.

### Pipeline

```text
SQL Error
    │
    ▼
LLM Repair
    │
    ▼
Execute Again
```

---

# 18. Safety Screener

## Definition

Checks whether an action is safe before execution.

Prevents destructive operations.

### Example

```text
DROP TABLE
     │
     ▼
Blocked
```

---

# 19. Tool Schema

## Definition

JSON specification describing a tool's:

- Name
- Inputs
- Outputs

Enables OpenAI Function Calling.

---

# 20. Tool Dispatch Map

## Definition

Maps tool names to Python functions.

Used by the agent to execute tool calls.

### Example

```python
{
    "execute_sql": execute_sql,
    "get_schema": get_schema
}
```

---

# 21. Self-Improvement Layer

## Definition

Agent learns from previous failures and evaluation feedback.

Improves future responses without changing the underlying model.

---

# 22. Verification

## Definition

Checks whether an answer is correct before returning it.

Often uses additional tools or evaluators.

---

# 23. Retry Mechanism

## Definition

Automatically retries generation after failures.

Often combined with evaluator feedback.

---

# 24. Reflection

## Definition

Agent critiques its own reasoning.

Corrects mistakes before producing the final answer.

---

# 25. Workflow Selection

| Workflow | Best Use Case |
|----------|---------------|
| Prompt Chaining | Multi-step tasks |
| Routing | Specialized tasks |
| Parallelization | Independent subtasks |
| Evaluator–Optimizer | High-quality outputs |
| Orchestrator–Worker | Large complex tasks |

---

# 26. Prompt Chaining vs Routing

| Prompt Chaining | Routing |
|-----------------|---------|
| Sequential | One path selected |
| Every step executes | Only selected workflow executes |
| Used for pipelines | Used for classification |

---

# 27. Parallelization vs Orchestrator

| Parallelization | Orchestrator |
|-----------------|--------------|
| Independent tasks | Dependent tasks |
| No coordinator logic | Central planner |
| Results merged equally | Planner controls execution |

---

# 28. Evaluator vs Optimizer

| Evaluator | Optimizer |
|------------|-----------|
| Finds mistakes | Fixes mistakes |
| Quality checker | Quality improver |
| No modification | Produces improved output |
