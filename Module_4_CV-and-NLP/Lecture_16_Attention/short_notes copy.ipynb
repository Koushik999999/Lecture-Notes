- Query (Q) = what we are searching/focusing for
- Key (K) = what each token offers/matches against
- Value (V) = information carried by the token (genereally K and V are same)
- If Q = K = V → self-attention
- If Q comes from one source and K,V from another → cross-attention
- Query = what we are focusing on (task-dependent; can be aspect word)
- Energy (score) = QK^T
- Attention weights = softmax(QK^T)
- Final attention output = softmax(QK^T)V
- This output is then passed to a task-specific layer (e.g., sigmoid for sentiment)
- Q, K, V are usually produced using trainable weight matrices

- Feed-forward network (FFN) is just a small neural network applied to each time step independently after attention.

- FFN adds non-linearity + feature transformation
