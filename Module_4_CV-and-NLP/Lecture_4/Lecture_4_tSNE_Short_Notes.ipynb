```python
def lucas_lehmer(p):
    if p == 2:
        return True
    
    M = (1 << p) - 1   # Faster than 2**p
    s = 4
    
    for _ in range(p - 2):
        s = (s * s - 2) % M
    
    return s == 0
```

# t-SNE (t-Distributed Stochastic Neighborhood Embedding)

distance betweeen points and probability of two points being close to each other is defined as follows:

- sigma(standard deviation) here is **NOT** the std of the entries set, written here as **sigma-i**
- sigma-i is the std of 15(hyper-parameter) nearest neighbors of point i
- 15 is called perplexity here
