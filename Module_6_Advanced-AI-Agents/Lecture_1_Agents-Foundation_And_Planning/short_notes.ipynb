# AI Agents Interview Cheat Sheet

---

# 1. AI Agent

## Definition
An AI agent is an LLM-powered system that can reason, plan, use tools, and take actions to achieve a goal.

Unlike a chatbot, it operates in a continuous:

**Perception → Reasoning → Action** loop.

---

# 2. Agent Loop

```text
User Goal
    │
    ▼
Reason / Plan
    │
    ▼
Choose Tool
    │
    ▼
Execute Tool
    │
    ▼
Observe Result
    │
    ▼
Continue or Finish
```

---

# 3. LLM vs AI Agent

| LLM | AI Agent |
|------|----------|
| Generates text | Achieves goals |
| Single response | Multi-step workflow |
| No actions | Can call tools |
| Stateless | Can maintain memory |
| No planning | Plans and reasons |

---

# 4. Deterministic Workflow

## Definition

Every step is predefined by the developer.

The same input always follows the same execution path.

### Example

```text
Login
  │
  ▼
Validate Password
  │
  ▼
Success / Failure
```

---

# 5. Agentic Workflow

## Definition

The LLM decides what to do next instead of following hardcoded logic.

Execution path changes dynamically based on the task.

### Example

```text
Question
    │
    ▼
Search Database?
    │
    ▼
Read PDF?
    │
    ▼
Call API?
    │
    ▼
Answer
```

---

# 6. Planning

## Definition

Breaking a complex task into smaller executable steps.

The LLM determines the sequence of actions before execution.

### Example

```text
Book Flight
    │
    ▼
Search Flights
    │
    ▼
Compare Prices
    │
    ▼
Book Ticket
```

---

# 7. Reasoning

## Definition

Logical thinking performed before deciding an action.

Helps the agent select the best next step.

---

# 8. ReAct (Reason + Act)

## Definition

Alternates between reasoning and tool execution.

**Think → Act → Observe → Think → Act**

### Example

```text
Thought
    │
    ▼
Search
    │
    ▼
Observation
    │
    ▼
Final Answer
```

---

# 9. Chain of Thought (CoT)

## Definition

Encourages the model to solve problems through intermediate reasoning steps.

Used mainly for reasoning, **not** tool use.

---

# 10. ReAct vs CoT

| Chain of Thought | ReAct |
|------------------|--------|
| Internal reasoning | Reasoning + Tool usage |
| No external actions | Calls APIs, databases, calculators |
| Ends with answer | Loops until task completion |

---

# 11. Persona

## Definition

Defines the agent's identity, expertise, behavior, and constraints.

Usually implemented as the **system prompt**.

### Example

```text
"You are a SQL expert."

"Never execute DELETE."

"Always verify schema first."
```

---

# 12. System Prompt

## Definition

Highest-priority instruction given to the LLM.

Defines role, rules, safety, and workflow.

---

# 13. Planner

## Definition

The component responsible for deciding what actions should happen next.

Usually, the LLM itself acts as the planner.

---

# 14. Executor

## Definition

Executes the action chosen by the planner.

Usually invokes tools or APIs.

---

# 15. Tool

## Definition

External function that the LLM can invoke.

Extends the model beyond text generation.

### Examples

- Calculator
- Weather API
- SQL Database
- Search Engine
- Python
- Email

---

# 16. Tool Calling (Function Calling)

## Definition

The LLM outputs structured JSON describing which function to call.

Your application executes the function and returns the result.

---

# 17. Function Schema

## Definition

JSON description of a tool including:

- Name
- Parameters
- Purpose

Helps the LLM decide when and how to call the function.

---

# 18. Tool Registry

## Definition

Dictionary or lookup table mapping tool names to implementations.

### Example

```python
tools = {
    "list_tables": list_tables,
    "execute_sql": execute_sql
}
```

---

# 19. Observation

## Definition

Output returned after a tool executes.

Used by the LLM to decide the next action.

---

# 20. Memory

## Definition

Allows an agent to remember previous interactions and results.

Prevents restarting from scratch every turn.

---

# 21. Short-Term Memory

## Definition

Current conversation and recent tool outputs.

Stored inside the context window.

---

# 22. Long-Term Memory

## Definition

Information stored externally across conversations.

Usually implemented using databases or vector stores.

---

# 23. Context Window

## Definition

Maximum number of tokens the model can process.

Limits how much memory fits in a prompt.

---

# 24. Tool Choice

## Definition

Decision of whether to answer directly or invoke a tool.

Modern LLMs learn this through function-calling fine-tuning.

---

# 25. MCP (Model Context Protocol)

## Definition

Open protocol for connecting LLMs to external tools and data sources.

Standardizes communication between models and applications.

---

# 26. Multi-Agent System

## Definition

Multiple specialized agents collaborate to solve one task.

Each agent has a dedicated responsibility.

### Example

```text
Supervisor
    │
    ▼
Research Agent
    │
    ▼
SQL Agent
    │
    ▼
Coding Agent
    │
    ▼
Final Response
```

---

# 27. Supervisor Agent

## Definition

Coordinates multiple specialized agents.

Assigns tasks and combines results.

---

# 28. Worker Agent

## Definition

Performs a specific task assigned by the supervisor.

### Examples

- SQL Agent
- Search Agent
- Coding Agent

---

# 29. Reflection

## Definition

Agent reviews its own previous answer and corrects mistakes.

Improves quality through self-evaluation.

---

# 30. Self-Consistency

## Definition

Generates multiple reasoning paths and chooses the most consistent answer.

Improves reasoning accuracy.

---

# 31. Task Decomposition

## Definition

Splits a large goal into smaller subtasks.

Essential for planning complex workflows.

---

# 32. State

## Definition

Current information available to the agent.

Includes:

- Conversation history
- Tool outputs
- Intermediate results

---

# 33. Goal

## Definition

Desired final outcome the agent is trying to achieve.

Drives planning and decision-making.

---

# 34. Environment

## Definition

External world the agent interacts with.

Examples include:

- APIs
- Databases
- Files
- Browsers
- Users

---

# 35. Action

## Definition

Operation performed by the agent.

### Examples

- Search
- Calculate
- Query database
- Send email

---

# 36. Observation

## Definition

Feedback received after executing an action.

Used to determine the next reasoning step.

---

# 37. Termination Condition

## Definition

Rule determining when the agent stops.

### Examples

- Answer found
- Task completed
- Maximum iterations reached

---

# 38. Human-in-the-Loop (HITL)

## Definition

Human approval required before sensitive actions.

Commonly used for:

- Finance
- Healthcare
- Legal applications

---

# 39. Agent Architecture

## Typical Architecture

```text
User
  │
  ▼
Planner (LLM)
  │
  ▼
Tool Selector
  │
  ▼
Tool Execution
  │
  ▼
Memory
  │
  ▼
Observation
  │
  ▼
Planner
  │
  ▼
Final Answer
```

---

# 40. AI Agent vs Workflow

| Workflow | AI Agent |
|----------|----------|
| Fixed logic | Dynamic reasoning |
| Hardcoded steps | LLM decides steps |
| No planning | Planning included |
| Limited flexibility | Adaptive |
| Predictable | Intelligent |
