- SILU is used instead of sigmoid

- SILU 
    - Sigmoid Linear Unit
    - Combines sigmoid and linear functions
    - Formula: f(x) = x * σ(x), where σ is the sigmoid function
    - Used in many modern architectures (e.g., LLaMA, GPT-3)
    - Computational cost is higher than ReLU, but provides better performance
