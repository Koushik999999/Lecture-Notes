```python
import pandas as pd
import unicodedata

# Load data
lat_long = pd.read_csv("India Cities LatLng.csv")
df = pd.read_csv("restaurant.csv")

# Normalize city names
lat_long = lat_long[['city', 'lat', 'lng']]
df['city'] = df['City'].str.lower()

def norm(word):
    return unicodedata.normalize('NFKD', word).encode('ascii', 'ignore').decode('utf-8')

lat_long['city'] = lat_long['city'].str.lower()
lat_long['city'] = lat_long['city'].apply(norm)
df['city'] = df['city'].apply(norm)



# Merge and clean
df = pd.merge(df, lat_long, on='city', how='left')
df.dropna(subset=['lat', 'lng'], inplace=True)

df.head()
```

```python
df['cuisine_list'] = df['Cuisine'].apply(lambda x: [c.strip() for c in x.split(",")])
mlb = MultiLabelBinarizer()
cuisine_binarized =pd.DataFrame(mlb.fit_transform(df['cuisine_list']), columns=mlb.classes_,index=df.index)

df = pd.concat([df.drop(columns=['cuisine_list']),cuisine_binarized],axis=1)
df.head()
df.drop(columns=['Cuisine','City'],inplace=True)
df.drop(columns=['city'],inplace=True)
```

```python
df.set_index("Restaurant",inplace=True)
```

```python
df.head()
```

```python
cosine_matrix = cosine_similarity(df)
```

```python
cosine_matrix
```

```python
cosine_df = pd.DataFrame(cosine_matrix, index=df.index, columns=df.index)
```

```python
cosine_df.head()
```

```python
def reccomend(restaurant,cosine_df,top_n):
    cos_values = cosine_df[restaurant].sort_values(ascending=False)
    return cos_values[1:top_n+1]

reccomend("Galaxy Surat",cosine_df,5)
```

```python
#Scale the longitude and latitude
df.info()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

scaled_df = scaler.fit_transform(df)

df_scaled = pd.DataFrame(scaled_df, index = df.index, columns=df.columns)
```

```python
df_scaled.head()
```

```python
cosine_matrix = cosine_similarity(df_scaled)
cosine_df = pd.DataFrame(cosine_matrix, index=df_scaled.index, columns=df_scaled.index)
def reccomend(restaurant,cosine_df,top_n):
    cos_values = cosine_df[restaurant].sort_values(ascending=False)
    return cos_values[1:top_n+1]

reccomend("Galaxy Surat",cosine_df,5)
```

```python
from sklearn.metrics.pairwise import manhattan_distances

manhattan_matrix = manhattan_distances(df_scaled)
manhattan_df = pd.DataFrame(manhattan_matrix, index = df_scaled.index, columns = df_scaled.index)
```

```python
def reccomend_manhattan(restaurant,top_n, manhattan_df):
    man_values = manhattan_df[restaurant].sort_values()
    return man_values[1:top_n+1]
```

```python
reccomend_manhattan("Galaxy Surat", 5, manhattan_df)
```

```python
import pandas as pd
df_fashion = pd.read_csv("fashion-mnist_test.csv")
```

```python
df_fashion.shape
```

```python
df_fashion = df_fashion.iloc[:, :-1]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
fashion_cosine_matrix = cosine_similarity(df_fashion)
fashion_cosine_df = pd.DataFrame(fashion_cosine_matrix, index=df_fashion.index, columns=df_fashion.index)
fashion_cosine_df.head()
```

```python
import numpy as np
ref_idx = 995
image_info = fashion_cosine_df.iloc[ref_idx].values

# Get top 10 similar images (excluding itself)
similar_images_idx = np.argsort(-image_info)[1:11]  # skip the image itself

print("Most similar image indices:", similar_images_idx)
```

```python
import matplotlib.pyplot as plt



plt.figure(figsize=(2, 2))
plt.title("Reference Image")
plt.imshow(df_fashion.iloc[ref_idx].values.reshape(28, 28), cmap='gray')
plt.axis('off')
plt.show()

# 7. Display top 10 similar images
plt.figure(figsize=(15, 3))
for i, idx in enumerate(similar_images_idx):
    plt.subplot(1, 10, i+1)
    plt.imshow(df_fashion.iloc[idx].values.reshape(28, 28), cmap='gray')
    plt.title(f"{idx}")
    plt.axis('off')
plt.suptitle("Top 10 Similar Images")
plt.show()
```
