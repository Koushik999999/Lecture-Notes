**Support Vector Machine**

--> Also called as maximum margin classifier

--> Not just the line, maximize the margin 


--> To do that maximize 2k/||w||

--> Minimize the loss fuinction

--> Loss function = ||w||/2

--> Hard Margin --> Not a single point goes inside the meargin

--> Because of this, line gets affected by only those points which are closest to it. 

--> Allow some error

Hinge loss = max(0, 1-y(wT . x))

We want to minimize ||w||/2 + C * sum(max(0, 1 - y_i * (w * x_i + b)))

If C is high, then model overfits and if C is low then model underfits

parametre --> optimized, learned

hyperparametre --> choose it and tune it accordingly

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

def tune_params(X,y, kernel_list, c_list): 
    X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=0, test_size=0.25)
    max_acc = 0
    for kernel in kernel_list:
        for c in c_list:
            model = SVC(kernel=kernel, C=c)
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            acc = accuracy_score(y_test, y_pred)
            if acc > max_acc:
                max_acc = acc
                best_params = {'kernel': kernel, 'C': c}
                best_support_vectors = len(model.support_vectors_)
    return best_params
            
```

```python
import numpy as np

weights = np.array([0.5,0.5])
inps = np.array([[0,0],[0,1],[1,0],[1,1]])
labels = np.array([0,0,0,1])
C = 1

# ||w|| = sqrt(w1^2 + w2^2)

norm = np.sqrt(np.sum(weights**2))
norm = norm/2

wx = np.dot(inps,weights)
wx = np.array(wx)
error = 0
for i in range(len(wx)):
    error += max(0,1-labels[i]*wx[i])

error = error/len(wx)
error = error*C + norm

np.round(error,3)
```
