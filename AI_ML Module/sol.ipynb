```python
class UserMainCode(object):
    @classmethod
    def minimumDistance(cls, P, X, L):
        min_abs = float('inf')

        for k in range(X + 1):
            # k is number of forward jumps, (X - k) is number of backward jumps
            pos = P + L * (2 * k - X)
            min_abs = min(min_abs, abs(pos))
        
        return min_abs
```

```python
def k_means_clustering(points: list[tuple[float, float]], k: int, initial_centroids: list[tuple[float, float]],max_iterations: int) -> list[tuple[float, float]]:
	n = len(initial_centroids)
	centroids = initial_centroids.copy()
	for i in range(max_iterations):
		updated_points : list[tuple[int, tuple[float, float]]] = []
		for point in points:
			dist: list[tuple[float, float]] = []
			for j in range(n):
				distance = (point[0] - initial_centroids[j][0])**2 + (point[1] - initial_centroids[j][1])**2
				dist.append((j, distance))
			min_j = min(dist, key=lambda x: x[1])[0]
			updated_points.append((min_j, point))
		final_centroids = []
		for j in range(n):
			sumx = 0
			sumy = 0
			count = 0
			for x in updated_points:
				if x[0] == j:
					sumx += x[1][0]
					sumy += x[1][1]
					count += 1
			if(count > 0):
				final_centroids.append((sumx / count, sumy / count))
			else:
				final_centroids.append(centroids[j])
		centroids = final_centroids
	return final_centroids
    
```

```python
points = [(1, 2), (1, 4), (1, 0), (10, 2), (10, 4), (10, 0)]
k = 2
initial_centroids = [(1, 1), (10, 1)]
max_iterations = 10
```

```python
k_means_clustering(points, k, initial_centroids, max_iterations)
```

```python
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import CountVectorizer

def cluster_tweets(tweets,k):
    vectorizer = CountVectorizer(lowercase=True,stop_words='english',ngram_range=(1,1))
    X = vectorizer.fit_transform(tweets)
    model = KMeans(n_clusters=k, random_state=42,n_init=10)
    model.fit(X)
    labels = model.labels_
    return labels
```

```python
tweets = ["Train is delayed", "Train is on time", "I love pizza", "Pizza is delicious"]
k = 2
```

```python
print(cluster_tweets(tweets,k))
```

```python
import numpy as np
def calc_r2(y_true, y_pred):
    
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
   
    
    #Calculate total sum of squares
    tot_ss = 0
    res_ss = 0
    for i in range(len(y_true)):
        tot_ss += (y_true[i] - y_pred[i])**2
        res_ss += (y_true[i] - np.mean(y_true))**2
    
    #Calculate residual sum of squares
     
    
    #Calculate r2
    r2 = 1 - tot_ss/res_ss
    
    return r2.round(2)
```
