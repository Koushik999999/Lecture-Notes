Given a Row wise and Column wise sorted matrix, find whether a given element is present in the matrix or not.

```python
def RowWiseAndColumnWise(arr,k):
    n = len(arr)
    m = len(arr[0])
    i = n-1
    j = 0
    while i>= 0 and j < m:
        if arr[i][j] == k:
            return True
        elif arr[i][j] > k:
            i -= 1
        else:
            j += 1
    return False
```

```python
a = [[-5, -2, 1, 13],
    [-4, 0, 3, 14],
    [-3, 2, 6, 18]]

print(RowWiseAndColumnWise(a,-5))
print(RowWiseAndColumnWise(a,14))
print(RowWiseAndColumnWise(a,8))
```

Given a list of intervals, merge all the overlapping intervals to produce a list that has only mutually exclusive intervals.

```python
def mergeIntervals(intervals):
    intervals.sort(key=lambda x: x[0])
    merged = []
    start = intervals[0][0]
    end = intervals[0][1]
    for i in range(1, len(intervals)):
        if(intervals[i][0] <= end):
            end = max(end, intervals[i][1])
        else:
            merged.append([start, end])
            start = intervals[i][0]
            end = intervals[i][1]
    merged.append([start, end])
    return merged
```
