# Evaluation Metrics for Machine Learning, NLP, and LLMs

---

# 1. Confusion Matrix

Everything starts with these four values.

| **Actual** | **Predicted** | **Meaning** |
|------------|---------------|-------------|
| Positive | Positive | True Positive (TP) |
| Negative | Negative | True Negative (TN) |
| Negative | Positive | False Positive (FP) |
| Positive | Negative | False Negative (FN) |

## True Positive (TP)

**Definition**

Model predicted **Positive**, and it actually was **Positive**.

**Example (Spam Detection)**

- **Actual:** Spam
- **Prediction:** Spam

✅ Correct

`TP += 1`

---

## True Negative (TN)

**Definition**

Model predicted **Negative**, and it actually was **Negative**.

**Example**

- **Actual:** Not Spam
- **Prediction:** Not Spam

✅ Correct

`TN += 1`

---

## False Positive (FP)

**Definition**

Model predicted **Positive**, but actually **Negative**.

Also called **Type-I Error**.

**Example**

- **Actual:** Not Spam
- **Prediction:** Spam

❌ Important email wrongly sent to spam.

`FP += 1`

---

## False Negative (FN)

**Definition**

Model predicted **Negative**, but actually **Positive**.

Also called **Type-II Error**.

**Example**

- **Actual:** Spam
- **Prediction:** Not Spam

❌ Spam reaches the inbox.

`FN += 1`

---

### Easy Way to Remember

|               | **Actual Positive** | **Actual Negative** |
|---------------|--------------------:|--------------------:|
| **Predicted Positive** | TP | FP |
| **Predicted Negative** | FN | TN |

---

# 2. Precision

## Formula

\[
\text{Precision}=\frac{TP}{TP+FP}
\]

**Meaning**

Among everything the model predicted as **Positive**, how many were actually correct?

**Interview Definition**

Precision measures how accurate the model's positive predictions are.

**Example**

- Predicted spam emails = **10**
- Actually spam = **8**

\[
\text{Precision}=\frac{8}{10}=0.8
\]

---

# 3. Recall

## Formula

\[
\text{Recall}=\frac{TP}{TP+FN}
\]

**Meaning**

Among all actual positives, how many did the model successfully find?

**Interview Definition**

Recall measures the model's ability to find all relevant positive instances.

**Example**

- Actual spam emails = **20**
- Detected = **18**

\[
\text{Recall}=\frac{18}{20}=0.9
\]

---

# 4. F1 Score

## Formula

\[
F1=2\times\frac{\text{Precision}\times\text{Recall}}
{\text{Precision}+\text{Recall}}
\]

**Meaning**

Harmonic mean of Precision and Recall.

Useful when both Precision and Recall are equally important.

---

# 5. Accuracy

## Formula

\[
\text{Accuracy}=
\frac{TP+TN}
{TP+TN+FP+FN}
\]

**Meaning**

Percentage of total predictions that are correct.

---

# 6. BLEU Score (Precision-based)

**BLEU = Bilingual Evaluation Understudy**

Used mainly for **Machine Translation**.

Measures how many generated **n-grams** appear in the reference sentence.

## Formula

\[
BLEU = BP \times
\exp\left(
\sum_{n=1}^{N}
w_n\log p_n
\right)
\]

where

- **BP** = Brevity Penalty
- \(p_n\) = Modified n-gram precision
- \(w_n\) = Weight (usually 1/4 for each of 1–4 grams)

### Modified Precision

\[
p_n=
\frac{\text{Matching n-grams}}
{\text{Generated n-grams}}
\]

### Brevity Penalty

If

\[
\text{Output Length}>\text{Reference Length}
\]

then

\[
BP=1
\]

otherwise

\[
BP=
e^{\left(
1-\frac{\text{Reference Length}}
{\text{Output Length}}
\right)}
\]

**Measures**

- ✅ Precision
- ❌ Not Recall

**Weakness**

Misses semantic similarity.

---

# 7. ROUGE Score (Recall-based)

**ROUGE = Recall-Oriented Understudy for Gisting Evaluation**

Mostly used for **Summarization**.

Measures how much of the reference is captured by the generated summary.

## Formula

\[
ROUGE=
\frac{\text{Matching Units}}
{\text{Reference Units}}
\]

**Measures**

- ✅ Recall

---

# 8. ROUGE-1

Uses **Unigrams (1-grams)**.

## Formula

\[
ROUGE\text{-}1=
\frac{\text{Matching Unigrams}}
{\text{Reference Unigrams}}
\]

---

# 9. ROUGE-2

Uses **Bigrams (2-grams)**.

## Formula

\[
ROUGE\text{-}2=
\frac{\text{Matching Bigrams}}
{\text{Reference Bigrams}}
\]

---

# 10. ROUGE-L

Uses the **Longest Common Subsequence (LCS)**.

## Formula

\[
ROUGE\text{-}L=
\frac{LCS}
{\text{Reference Length}}
\]

Captures

- Sentence structure
- Word ordering

---

# 11. BERTScore

Instead of exact word matching, BERTScore uses **BERT embeddings**.

Measures **semantic similarity**.

### Example

**Reference**

> The cat sat on the mat.

**Generated**

> A kitten rested on the rug.

- BLEU → Low
- BERTScore → High

because the meanings are similar.

### Conceptual Formula

For every generated token,

- Find the highest cosine similarity with a reference token embedding.
- Average all similarities.

\[
\text{BERTScore}
=
\text{Average Cosine Similarity of Contextual Embeddings}
\]

---

# 12. METEOR

Uses

- Precision
- Recall
- Synonyms
- Stemming

Better correlated with human judgment than BLEU.

---

# 13. CIDEr

Used for **Image Captioning**.

Measures similarity using **TF-IDF weighted n-grams**.

---

# 14. Exact Match (EM)

## Formula

\[
EM=
\begin{cases}
1,& \text{if prediction exactly matches reference}\\
0,& \text{otherwise}
\end{cases}
\]

Used in

- Question Answering
- SQuAD

---

# 15. Perplexity (PPL)

Measures how surprised the language model is.

Lower is better.

## Formula

\[
PPL=
\exp\left(
-\frac{1}{N}
\sum
\log P(w_i)
\right)
\]

---

# 16. Human Evaluation

Usually scored on

- Fluency
- Correctness
- Relevance
- Helpfulness
- Factuality
- Coherence

Gold standard for evaluating LLMs.

---

# 17. Hallucination Rate

Measures how often the model generates unsupported or false information.

Lower is better.

---

# 18. Faithfulness

Measures whether the generated answer is supported by the provided context.

Critical for

- Retrieval-Augmented Generation (RAG)
- Summarization

---

# 19. Answer Relevancy

Measures whether the answer actually addresses the user's question.

Higher is better.

---

# 20. Context Precision (RAG)

Measures how much of the retrieved context is actually relevant.

Higher means better retrieval quality.

---

# 21. Context Recall (RAG)

Measures how much of the necessary information was successfully retrieved.

Higher means fewer missing facts.

---

# 22. Groundedness

Measures whether every generated statement can be traced back to the supplied documents or evidence.

Very important in production Retrieval-Augmented Generation (RAG) systems.
